```python
print("PROGRAM SPLTV")
print("METODE ELIMINASI GAUSS")
print("======================")

import numpy as np

A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

B = np.array([8, -11, -3], dtype=float)

M = np.column_stack((A, B))

print("Matriks Augmented:")
print(M)

n = len(B)

for i in range(n):
    pivot = i + np.argmax(np.abs(M[i:, i]))
    M[[i, pivot]] = M[[pivot, i]]

    if abs(M[i, i]) < 1e-10:
        raise ValueError("Sistem tidak memiliki solusi tunggal")

    for j in range(i + 1, n):
        faktor = M[j, i] / M[i, i]
        M[j] = M[j] - faktor * M[i]

x = np.zeros(n)

for i in range(n - 1, -1, -1):
    x[i] = (
        M[i, n] - np.dot(M[i, i+1:n], x[i+1:n])
    ) / M[i, i]

print("\nHASIL PERHITUNGAN:")
print("Nilai x =", x[0])
print("Nilai y =", x[1])
print("Nilai z =", x[2])
```

1. Program ini digunakan untuk menyelesaikan Sistem Persamaan Linear Tiga Variabel (SPLTV) menggunakan metode Eliminasi Gauss. Program menggunakan bahasa Python dan library NumPy untuk melakukan perhitungan matriks. Hasil perhitungan menunjukkan bahwa nilai x = 2, y = 3, dan z = -1.

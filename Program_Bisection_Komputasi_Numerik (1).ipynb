```python
print("PROGRAM METODE BISECTION")
print("=========================")

# Fungsi persamaan
def f(x):
    return x**3 - x - 2

# Nilai awal
a = 1.0
b = 2.0

toleransi = 0.000001
maks_iterasi = 100

# Memeriksa interval
if f(a) * f(b) >= 0:
    print("Interval tidak memenuhi syarat")
else:
    for i in range(1, maks_iterasi + 1):
        c = (a + b) / 2

        print("Iterasi:", i, "Nilai c:", c)

        if abs(f(c)) < toleransi:
            break

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("\nHASIL PERHITUNGAN")
    print("Akar persamaan =", c)
    print("Nilai f(c) =", f(c))
```

Penjelasan Program Metode Bisection

Program ini digunakan untuk mencari akar persamaan nonlinear menggunakan metode Bisection. Persamaan yang digunakan adalah (f(x)=x^3-x-2), dengan interval awal a = 1 dan b = 2.

Program membagi interval menjadi dua bagian dengan menghitung titik tengah. Proses tersebut dilakukan berulang kali sampai memenuhi nilai toleransi yang ditentukan.

Hasil perhitungan menunjukkan akar persamaan sebesar 1.5213797092437744 dengan jumlah 22 iterasi.
